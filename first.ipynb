```python
df.rename(
    columns={
        "TWF":"Tool_Wear_Failure",
        "HDF":"Heat_Dissipation_Failure",
        "PWF":"Power_Failure",
        "OSF":"Overstrain_Failure",
        "RNF":"Random_Failure"
    },
    inplace=True
)
```

```python


from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier, plot_tree

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)
```

```python
features = [
    "Type",
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]",
    "Tool_Wear_Failure",
    "Heat_Dissipation_Failure",
    "Power_Failure",
    "Overstrain_Failure",
    "Random_Failure"
]

target = "Machine failure"

X = df[features]
y = df[target]
```

```python
print(X.columns.tolist())
print(y.name)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True))
```

```python
categorical_features = ["Type"]

numeric_features = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]",
    "Tool_Wear_Failure",
    "Heat_Dissipation_Failure",
    "Power_Failure",
    "Overstrain_Failure",
    "Random_Failure"
]

try:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

preprocessor = ColumnTransformer([
    (
        "categorical",
        encoder,
        categorical_features
    ),
    (
        "numeric",
        "passthrough",
        numeric_features
    )
])
```

```python
model = Pipeline([
    (
        "preprocessing",
        preprocessor
    ),
    (
        "classifier",
        DecisionTreeClassifier(
            max_depth=5,
            min_samples_split=10,
            min_samples_leaf=5,
            criterion="gini",
            random_state=42
        )
    )
])

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_probability = model.predict_proba(X_test)[:, 1]
```

```python
prediction_table = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred,
    "Failure_Probability": y_probability
})

print(prediction_table.head(20))
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_probability)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-score :", f1)
print("ROC-AUC  :", roc_auc)
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["No Failure", "Failure"]
    )
)
```

```python
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline

svc_model = Pipeline([
    ("preprocessing", preprocessor),
    (
        "classifier",
        SVC(
            kernel="rbf",
            C=1.0,
            gamma="scale",
            probability=True,
            random_state=42
        )
    )
])

svc_model.fit(X_train, y_train)
```

```python
y_pred_svc = svc_model.predict(X_test)
y_prob_svc = svc_model.predict_proba(X_test)[:, 1]

print("Accuracy:", accuracy_score(y_test, y_pred_svc))
print("ROC-AUC:", roc_auc_score(y_test, y_prob_svc))

print(
    classification_report(
        y_test,
        y_pred_svc,
        target_names=["No Failure", "Failure"]
    )
)
```

```python
from sklearn.ensemble import VotingClassifier

ensemble_model = VotingClassifier(
    estimators=[
        ("random_forest", rf_model),
        ("gradient_boosting", gb_model),
        ("ffnn", ffnn_model),
        ("svc", svc_model)
    ],
    voting="soft",
    weights=[3, 3, 2, 2]
)

ensemble_model.fit(X_train, y_train)
```

```python
y_pred_ensemble = ensemble_model.predict(X_test)
y_prob_ensemble = ensemble_model.predict_proba(X_test)[:, 1]

print("Accuracy:", accuracy_score(y_test, y_pred_ensemble))
print("ROC-AUC:", roc_auc_score(y_test, y_prob_ensemble))

print(
    classification_report(
        y_test,
        y_pred_ensemble,
        target_names=["No Failure", "Failure"]
    )
)
```

```python
model_results = []

trained_models = {
    "Random Forest": rf_model,
    "Gradient Boosting": gb_model,
    "FFNN": ffnn_model,
    "SVC": svc_model,
    "Ensemble": ensemble_model
}

for name, trained_model in trained_models.items():
    predictions = trained_model.predict(X_test)
    probabilities = trained_model.predict_proba(X_test)[:, 1]

    model_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions),
        "Recall": recall_score(y_test, predictions),
        "F1": f1_score(y_test, predictions),
        "ROC-AUC": roc_auc_score(y_test, probabilities)
    })

comparison_df = pd.DataFrame(model_results)

comparison_df.sort_values(
    by="ROC-AUC",
    ascending=False
)
```

```python
comparison_df.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1", "ROC-AUC"]
].plot(
    kind="bar",
    figsize=(12, 6)
)

plt.ylim(0, 1)
plt.ylabel("Score")
plt.title("Model Performance Comparison")
plt.xticks(rotation=45)
plt.legend(loc="lower right")
plt.show()
```
